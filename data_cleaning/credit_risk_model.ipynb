```python
df_customers = pd.read_csv('/content/customers.csv')
df_loans = pd.read_csv('/content/bureau_data.csv')
df_bureau = pd.read_csv('/content/loans.csv')
```

```python
df = pd.merge(df_customers, df_loans, on='cust_id')
df = pd.merge(df, df_bureau, on='cust_id')
df.head(3)
```

```python
df.info()
```

```python
df['default'] = df['default'].astype(int)
df.default.value_counts()
```

##### Due to the Data Leakage problem and Target Leakage problem, we are first split the data set.

```python
X = df.drop('default', axis='columns')
y = df['default']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)

df_train = pd.concat([X_train, y_train], axis='columns')
df_test = pd.concat([X_test, y_test], axis='columns')

df_train.head()
```
